```python
import statistics

class_marks=[65, 72, 80, 55, 90, 68, 75, 82, 60, 78]

mean_mark = statistics.mean(class_marks)
median_mark = statistics.median(class_marks)
std_mark = statistics.pstdev(class_marks)

print("Mean:", mean_mark)
print("Median:", median_mark)
print("Population Standard Deviation:", std_mark)
```
